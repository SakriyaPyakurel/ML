# Perceptron loss function 

```python
# importing necessary libraries 
from sklearn.datasets import make_classification 
import numpy as np 
import matplotlib.pyplot as plt 
```

```python
# creating and showcasing the data visually 
X,y = make_classification(n_samples=100,n_features=2,n_informative=1,n_redundant=0,
                          n_classes=2,n_clusters_per_class=1,random_state=41,hypercube=False,class_sep=15)
plt.figure(figsize=(10,6)) 
plt.scatter(X[:,0],X[:,1],c=y,cmap='winter',s=100)
```

Perceptron function definition with loss function combination

```python
def perceptron(X,y,learning_rate=0.1):
    w1=w2=b=1 
    for i in range(1000):
        for i in range(X.shape[0]):
            # checking condition 
            z = w1*X[i][0]+w2*X[i][1]+b
            if z*y[i] < 0:
                w1 = w1 + learning_rate*y[i]*X[i][0] 
                w2 = w2 + learning_rate*y[i]*X[i][1] 
                b = b + learning_rate*y[i] 
    return w1,w2,b
```

```python
# determining the value of w1,w2 and b through perceptron function 
w1,w2,b = perceptron(X,y)
print(w1) 
print(w2) 
print(b)
```

```python
# determining slope(m) and intercept(c) 
m = -(w1/w2) 
c = -(b/w2) 
print(m,c) 
```

```python
x_input = np.linspace(-3,3,100) 
y_input = m*x_input+c 
plt.figure(figsize=(10,6)) 
plt.plot(x_input,y_input,color='red',linewidth=5) 
plt.scatter(X[:,0],X[:,1],c=y,cmap='winter',s=100)
plt.ylim(-5,4)
```

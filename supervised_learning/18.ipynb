## Training a custom perceptron model from scratch

creating a custom linear data

```python
# importing necessary libraries 
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification

# Generating linear data 
X, y = make_classification(
    n_samples=100,
    n_features=2,
    n_informative=1,
    n_redundant=0,
    n_classes=2,
    n_clusters_per_class=1,
    random_state=41,
    hypercube=False,
    class_sep=10
)
```

```python
# plotting the created data 
plt.figure(figsize=(10,6)) 
plt.scatter(X[:,0],X[:,1],c=y,cmap='winter',s=100)
```

```python
# utility step function definition 
def step(z):
    return 1 if z>0 else 0
```

Perceptron function definition

```python
# Perceptron function 
def perceptron(X,y,learning_rate=0.1):
    X = np.insert(X,0,1,axis=1) 
    weights = np.ones(X.shape[1]) 

    for i in range(1000):
        j = np.random.randint(0,100)
        y_cap = step(np.dot(X[j],weights))
        weights = weights + learning_rate*(y[j]-y_cap)*X[j] 
    return weights[0],weights[1:]
```

```python
# determining intercept and coefficients 
intercept_,coef_ = perceptron(X,y)
print(coef_) 
print(intercept_)
```

```python
# determining slope(m) and bias(b) 
m = -(coef_[0]/coef_[1]) 
b = -(intercept_/coef_[1])
print(m) 
print(b)
```

Perceptron line plot

```python
x_input = np.linspace(-3,3,100) 
y_input = m*x_input+b
plt.figure(figsize=(10,6)) 
plt.plot(x_input,y_input,color='brown',linewidth=3) 
plt.scatter(X[:,0],X[:,1],c=y,cmap='winter',s=100) 
plt.ylim(-4,2)
```

Successfully trained a perceptron model for seggregating linear data into perfect binary clusters.

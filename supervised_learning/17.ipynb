## Perceptron

An supervised learning algorithm and simplest type of artificial neural network used for binary classification.

Trying perceptron on linear data

```python
# importing necessary libraries 
import pandas as pd
import seaborn as sb 
from sklearn.linear_model import Perceptron
```

```python
# importing linear data
df = pd.read_csv('linear_placement_data.csv') 
print(df.shape)  
df.head(15)
```

```python
#visualizing the imported data in scatter plot
sb.scatterplot(x=df['cgpa'],y=df['resume_score'],hue=df['placed'])
```

```python
# determining x and y 
X_linear = df.iloc[:,0:2] 
Y_linear = df.iloc[:,-1]
```

```python
# initiliazing perceptron model object and training it
p_linear = Perceptron() 
p_linear.fit(X_linear,Y_linear) 
```

```python
# determining coefficents(w1 and w2)
w1,w2 = p_linear.coef_[0]
print(f"w1:{w1}\nw2:{w2}")
```

```python
# determining intercept or bias(b) 
print(f"bias(b): {p_linear.intercept_[0]} ")
```

```python
# for checking function is successfully plotted or not 
from mlxtend.plotting import plot_decision_regions 
plot_decision_regions(X_linear.values,Y_linear.values,clf=p_linear,legend=2)
```

Trying perceptron on non linear data

```python
# importing non linear data 
df = pd.read_csv('nonlinear_placement_data.csv') 
print(df.shape) 
df.head(15)
```

```python
# visualizing the imported data in a scatter plot 
sb.scatterplot(x=df['cgpa'],y=df['resume_score'],hue=df['placed'])
```

```python
# determining x and y 
X_non_linear = df.iloc[:,0:2] 
Y_non_linear = df.iloc[:,-1]
```

```python
# initiliazing perceptron model object and training it
p_non_linear = Perceptron() 
p_non_linear.fit(X_non_linear,Y_non_linear) 
```

```python
# determining coefficents(w1 and w2)
w1,w2 = p_non_linear.coef_[0]
print(f"w1:{w1}\nw2:{w2}")
```

```python
# determining intercept or bias(b) 
print(f"bias(b): {p_non_linear.intercept_[0]} ")
```

```python
# for checking function is successfully plotted or not 
from mlxtend.plotting import plot_decision_regions 
plot_decision_regions(X_non_linear.values,Y_non_linear.values,clf=p_non_linear,legend=2)
```

Comparison between perceptron used on linear and non linear data 

```python
# comparing scores of both perceptron models 
print(f"Perceptron score on linear data: {p_linear.score(X_linear,Y_linear)}\nPerceptron score on non linear data: {p_non_linear.score(X_non_linear,Y_non_linear)}")
```

## Conclusion

Perceptron works on linear data or slightly non linear data more properly than completely non linear data.
